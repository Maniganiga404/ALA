```python
def test_compare_handles():
  h1 = create_handle("9591", "hdfc")
  h2 = create_handle("3456", "hdfc")
  assert not compare_handles(h1, h2)
  assert compare_handles(h1, h1)
  assert not compare_handles(h2, h1)
  assert compare_handles(h2, h2)

test_compare_handles()
```
