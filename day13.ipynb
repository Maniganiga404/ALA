```python
#lambda
#Write a lambda function called cube that takes a number and returns its cube
cube = lambda a : a*a*a
print(cube(2))
```

```python
#map
#multiply by 3
numbers = [2, 4, 6, 8, 10]
result = list(map(lambda n: n * 3, numbers))
print(result)
```

```python
#filter
numbers = [10, 15, 20, 25, 30, 35, 40]
result = list(filter(lambda n : n > 25, numbers))
print(result)
```

```python
#reduce
#product
from functools import reduce

numbers = [2, 3, 4, 5]
product = reduce(lambda n,m : n * m,numbers)
print(product)
```

```python
#sorted with key
names = ["Manish", "Raj", "Bhavish", "Om", "Kiran"]

result = sorted(names, key = lambda name : len(name))

print(result)
```

```python
#Use filter() to get numbers greater than 10
#Use map() to square those numbers
#Use reduce() to find their total

from functools import reduce

numbers = [12, 5, 8, 21, 30, 7, 16, 3]

result = list(filter(lambda n : n >10,numbers))

print(result)

square = list(map(lambda n : n*n, result))

print(square)

total = reduce(lambda a,b : a+b, square)

print(total)
```

```python
#any and all

numbers = [10, 20, 30, 40, 50]

result = any(n > 45 for n in numbers)

print(result)

result1 = all(n>5 for n in numbers)

print(result1)
```
